```python
import pandas as pd
from pathlib import Path

special_days = {
    1: "آغاز حمله مغول به ایران در پاییز 598 خورشیدی",
    5: "روز جهانی جهانگردی [ September 27 ]",
    7: "روز آتش‌نشانی و ایمنی | روز بزرگداشت شمس تبریزی",
    8: "روز بزرگداشت مولوی | روز جهانی ناشنوایان [ September 30 ] | روز جهانی ترجمه و مترجم [ September 30 ]",
    9: "روز جهانی سالمندان [ October 1 ]",
    10: "مهر روز، جشن مهرگان",
    12: "آغاز هفته جهانی فضا [ October 4 ]",
    13: "روز جهانی معلم [ October 5 ]",
    14: "روز دامپزشکی",
    16: "روز ملی کودک",
    17: "روز جهانی پست [ October 9 ]",
    18: "روز جهانی مبارزه با حکم اعدام [ October 10 ]",
    19: "روز جهانی دختر [ October 11 ]",
    20: "روز بزرگداشت حافظ",
    21: "روز پیروزی کاوه و فریدون بر ضحاک",
    22: "روز جهانی استاندارد [ October 14 ]",
    23: "روز جهانی عصای سفید [ October 15 ]",
    24: "روز جهانی غذا [ October 16 ]",
    25: "روز جهانی ریشه‌کنی فقر [ October 17 ]",
    26: "روز تربیت بدنی و ورزش",
    28: "زادروز ستارخان ملقب به سردار ملی و از سرداران جنبش مشروطه ایران",
    29: "روز ملی کوهنورد"
}

weekdays = {
    1: "چهارشنبه",
    2: "پنجشنبه",
    3: "جمعه",
    4: "شنبه",
    5: "یک شنبه",
    6: "دوشنبه",
    7: "سه شنبه",
}

en_year = "1405"
fa_year = "۱۴۰۵"
english_month_name = "mehr"
persian_month_name = "مهر"
month_number = 7
start_id = 8081
days_in_month = 30

BASE_DIR = Path("ExcelFiles") / "2585"
excel_name = f"07-images-{english_month_name}-{en_year}.xlsx"
# csv_name = f"images-{english_month_name}-{en_year}.csv"s
EXCEL_FILE = BASE_DIR / excel_name
```

```python
def to_persian_number(n):
    mapping = {"0":"۰","1":"۱","2":"۲","3":"۳","4":"۴","5":"۵","6":"۶","7":"۷","8":"۸","9":"۹"}
    return "".join(mapping.get(ch, ch) for ch in str(n))

def weekday_for(day):
    return weekdays[((day - 1) % 7) + 1]

normal_description_template = f"""قیمت روز طلا و سکه امروز {{weekday}} {{day}} {persian_month_name} {fa_year} را در وب‌سایت طلای شهریور ببینید.
این صفحه شامل نرخ لحظه‌ای طلا، سکه، شمش و ارز دیجیتال است و هر روز دو بار، ساعت‌های ۱۲:۱۲ و ۱۷:۱۷ بروزرسانی می‌شود."""

rows = []
times = ["۱۲:۱۲", "۱۷:۱۷"]

ids = list(range(start_id, start_id + days_in_month * 2))
id_index = 0

for day in range(1, days_in_month + 1):
    weekday = weekday_for(day)
    pday = to_persian_number(day)

    for time in times:
        current_id = ids[id_index]
        id_index += 1

        if day in special_days:
            reason = special_days[day]
            alt_text = (
                f"قیمت طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} ساعت {time} – {reason} | طلای شهریور"
            )
            title = (
                f"قیمت طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} ساعت {time} – {reason}"
            )
            excerpt = (
                f"آخرین قیمت روز طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} در طلای شهریور – بروزرسانی {time}"
            )
            description = (
                f"قیمت روز طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} را در وب‌سایت طلای شهریور ببینید.\n"
                f"امروز {reason} است.\n"
                "در این صفحه نرخ لحظه‌ای طلا، سکه، شمش و ارز دیجیتال را دنبال کنید."
            )
        else:
            alt_text = (
                f"قیمت طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} ساعت {time} | طلای شهریور"
            )
            title = (
                f"قیمت طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} ساعت {time}"
            )
            excerpt = (
                f"آخرین قیمت روز طلا و سکه امروز {weekday} {pday} {persian_month_name} {fa_year} در طلای شهریور – بروزرسانی {time}"
            )
            description = normal_description_template.format(weekday=weekday, day=pday)

        rows.append({
            "Id_pic": current_id,
            "Alt Text": alt_text,
            "Title": title,
            "Excerpt": excerpt,
            "Description": description
        })

df = pd.DataFrame(rows)

df.to_excel(EXCEL_FILE, index=False)
# df.to_csv(csv_name, index=False, encoding="utf-8-sig")

df.head()
```
